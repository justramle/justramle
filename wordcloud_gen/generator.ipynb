```python
import random
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# Soft pastel rainbow palette (blue, mint, pink, peach, lavender)
HUES = [210, 165, 340, 25, 270, 190]

# Boba types get extra weight so every kind of boba reads clearly, not squeezed tiny
BOBA_WORDS = {
    "TraSua", "BrownSugarBoba", "TaroMilkTea", "MatchaMilkTea", "ThaiMilkTea",
    "WinterMelonTea", "OolongMilkTea", "JasmineMilkTea", "HoneyBlackTea",
    "CoconutMilkTea", "MangoGreenTea", "StrawberryMilkTea", "OkinawaMilkTea",
    "TigerSugarMilk", "CheeseFoamTea", "TapiocaPearls", "PoppingBoba",
    "CrystalBoba", "GrassJellyBoba", "AloeVeraBoba", "PuddingMilkTea",
    "RedBeanMilkTea", "LycheeJellyTea", "CoconutJelly", "NataDeCoco",
    "HongKongMilkTea", "RoastedMilkTea", "PeachOolongTea",
}
BOBA_BOOST = 3

# A handful of iconic Vietnamese dishes get the same boost so food stands out next to the boba
ICONIC_FOOD_WORDS = {
    "PhoBo", "BanhMi", "BunCha", "GoiCuon", "BanhXeo", "ComTam", "BunBoHue",
    "Che", "CaPheSuaDa", "BanhCuon", "ChaGio", "MiQuang", "SinhTo",
}
FOOD_BOOST = 3

def pastel_palette(word, font_size, position, orientation, random_state=None, **kwargs):
    hue = random.choice(HUES)
    sat = random.randint(50, 65)
    light = random.randint(50, 62)
    return "hsl(%d, %d%%, %d%%)" % (hue, sat, light)

def plot_word_cloud():
    # Read file from txt
    with open('worddoc.txt', 'r', encoding='UTF-8') as file:
        words = [line.strip() for line in file if line.strip()]

    # Boost both boba types and iconic Vietnamese dishes so both stand out together
    boosted = []
    for w in words:
        boosted.append(w)
        if w in BOBA_WORDS:
            boosted.extend([w] * (BOBA_BOOST - 1))
        elif w in ICONIC_FOOD_WORDS:
            boosted.extend([w] * (FOOD_BOOST - 1))
    wordlist = ' '.join(boosted)

    # Wide rectangle, transparent background (no box), fully horizontal
    wordcloud = WordCloud(width=1500, height=650,
                          background_color=None, mode='RGBA',
                          font_path='Montserrat-Bold.otf',
                          color_func=pastel_palette,
                          collocations=False,
                          prefer_horizontal=1.0,
                          margin=6,
                          max_words=160,
                          max_font_size=140,
                          relative_scaling=0.5,
                          random_state=42,
                          ).generate(wordlist)

    plt.figure(figsize=(15, 6.5))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis('off')
    plt.savefig('../images/boba_wordcloud.png', bbox_inches="tight", pad_inches=0.2, transparent=True, dpi=200)
    plt.show()


plot_word_cloud()
```
